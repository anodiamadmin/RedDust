```python
from pydantic import BaseModel, EmailStr
from typing import List, Dict, Optional


class PatientData(BaseModel):
    name: str
    email: Optional[EmailStr] = None
    age: int
    weight: float
    married: bool
    allergies: Optional[List[str]] = None
    contact_info: dict[str, str]


def add_patient_data(patient: PatientData):
    print(f"\nAdding patient data for {patient.name}...")
    print(f"Patient Name: {patient.name}, Age: {patient.age}, Weight: {patient.weight} kg")
    print(f"Married: {patient.married}, Allergies: {patient.allergies}")
    print(f"Contact Info: Phone - {patient.contact_info.get('phone')}, Email - {patient.email}")
    print("Patient data added successfully!")


# def update_patient_data(patient: PatientData):
#     print(f"\nUpdating patient data for {patient.name}...")
#     print(f"Patient Name: {patient.name}, Age: {patient.age}, Weight: {patient.weight} kg")
#     print("Patient data updated successfully!")


# patient1_data = {"name": "John Doe", "age": 30, "weight": 70.5, "married": True, "allergies": ["penicillin"], "contact_info": {"phone": "123-456-7890"}}
patient2_data = {"name": "Jane Smith", "age": 25, "weight": 60.0, "married": False, "contact_info": {"phone": "098-765-4321"}, "email": "abc@eg.com"}

# patient1 = PatientData(**patient1_data)
patient2 = PatientData(**patient2_data)

# add_patient_data(patient1)
# update_patient_data(patient1)
add_patient_data(patient2)
# update_patient_data(patient2)
```
